```python
def order_pizza(pizza_count):
    if pizza_count < 1:
        return 'Error message: "Invalid Order"'
    elif 1 <= pizza_count <= 10:
        return 'Success message'
    elif 11 <= pizza_count <= 99:
        return 'Error message: "Only 10 Pizza can be ordered"'
    elif pizza_count >= 100:
        return 'Error message: "Invalid Order"'
```

```python
df['Actual Result'] = df['Order Value'].apply(order_pizza)

df['Result Matches'] = df['Actual Result'] == df['Expected Result']
```
