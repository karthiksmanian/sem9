```python
def calculate_discount(purchase_value):
    if purchase_value <= 50:
        return 'No discount'
    elif 50 < purchase_value <= 200:
        return '5% discount'
    elif 200 < purchase_value <= 500:
        return '10% discount'
    else:
        return '15% discount'
```

```python
df['Actual Result'] = df['Purchase Value'].apply(calculate_discount)
df['Result Matches'] = df['Actual Result'] == df['Expected Result']
```
